```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = r'Q:\Documents\Projets\Methanogens'
df1 = pd.read_excel(path + r'\Metadata_methano_VM.xlsx', sheet_name='Feuil1', header=0, index_col=0)
df2 = pd.read_excel(path + r'\Methanogens_proteins.xlsx', sheet_name='Sheet1', header=0, index_col=0)

merged_df = pd.merge(df1, df2, left_on=['Species'], right_on=['organism'], how='inner')
# merged_df.to_excel(path + r'\Merged_Metadata_proteins.xlsx', index=False)
```

```python
for column in ['Max growth T', 'Optimal growth T']:
    merged_df[column] = pd.to_numeric(merged_df[column], errors='coerce')
    sns.scatterplot(data = merged_df, x='ZC', y=column, hue='Class', palette='Set1')
    plt.legend(loc=(1,0))
    plt.show()
```
